```python
# Normal function: “Here’s all the data, take it!”


def add():
    return [1, 2, 3]

print(add())
```

```python
# Generator function: “I’ll give you one value at a time when you ask for it.”
def add():
    yield 1
    yield 2
    yield 3
x=add()
print(add())
```

```python
print(next(x))
print(next(x))
print(next(x))
```
